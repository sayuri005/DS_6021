```python
import pandas as pd 
import numpy as np
import plotly.express as px 
```

```python
#1 Used AI 
data = pd.read_csv("WA_Marketing-Campaign.csv")
data["Promotion"] = data["Promotion"].astype(str)
fig = px.bar(
    data,
    x="LocationID",              # your store column
    y="SalesInThousands",
    color="Promotion",           # bars colored by promotion
    barmode="group",             # side-by-side; use "stack" to stack them
    title="Sales by Store and Promotion"
)
fig.update_xaxes(type="category")   # treat store IDs as discrete categories
fig.show()
```
