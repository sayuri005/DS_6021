3.Stores using promotion 1 made around 10.77 thousands of dollars more weekly than stores using promotion 2.

```python
#4
prom1_sales = (
    data.loc[
        data["Promotion"] == "1",
        "SalesInThousands"
    ]
    .to_numpy()
)

prom2_sales = (
    data.loc[
        data["Promotion"] == "2",
        "SalesInThousands"
    ]
    .to_numpy()
)
my_seed = np.random.default_rng(500)
B = 5000

boot_spend_diff = np.empty(B)

for b in range(B):

    promotion1 = my_seed.choice(
        prom1_sales,
        size=len(prom1_sales),
        replace=True
    )

    promotion2 = my_seed.choice(
        prom2_sales,
        size=len(prom2_sales),
        replace=True
    )

    boot_spend_diff[b] = (
        promotion2.mean()
        - promotion1.mean()
    )
lower, upper = np.percentile(
    boot_spend_diff,
    [2.5, 97.5]
)
print(lower, upper)
```

2.After ANOVA shows that there is a significant difference, one can decide which two to compare by referring back to the explanatory analysis.For example, there seems to be differnce between promotion 3 and 2, so I would do a boostrap condifence interval to comapare.


3.When there are seveal comparisons, the probability to see significant differences increase becasue one is doing more comparisons and thus increases the probability that one sees "significant" differences when it was actually simply due to chance. 

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
model_sales = ols(
    "SalesInThousands ~ C(Promotion)",
    data=data
).fit()
anova_table = sm.stats.anova_lm(model_sales, typ=2)

anova_table
```
